## 3. PyTorch 텐서 연산 확인

PyTorch가 GPU를 쓸 수 있는지도 같이 본다. Mac이면 MPS(Metal), 리눅스 GPU면 CUDA.
없으면 CPU만 써도 Layer 1~2는 충분하다.

```python
# 디바이스 고르기
if torch.cuda.is_available():
    device = 'cuda'
elif torch.backends.mps.is_available():
    device = 'mps'
else:
    device = 'cpu'
print(f'사용 가능 디바이스: {device}')

# 작은 텐서 연산
x = torch.randn(3, 4, device=device)
y = torch.randn(4, 5, device=device)
z = x @ y
print(f'x shape: {x.shape}, y shape: {y.shape}, z = x @ y shape: {z.shape}')
print('OK!')
```

## 4. sklearn 내장 의료 데이터 로딩 확인

```python
from sklearn.datasets import load_breast_cancer, load_diabetes

bc = load_breast_cancer()
db = load_diabetes()
print(f'Breast cancer: X shape {bc.data.shape}, y shape {bc.target.shape}')
print(f'  features: {bc.feature_names[:5]} ...')
print(f'Diabetes    : X shape {db.data.shape}, y shape {db.target.shape}')
print(f'  features: {db.feature_names}')
```
